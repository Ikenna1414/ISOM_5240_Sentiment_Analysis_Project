## Run candidate experiments
The unchanged FinBERT reference does not need three stochastic training runs. Fine-tuned candidates use seeds 42, 43 and 44. Test data is never used for selection.


```python
for seed in [42, 43, 44]:
    command=[sys.executable,"-m","finsignal.train","--data",str(data),"--task","sentiment","--model","distilbert/distilbert-base-uncased","--seed",str(seed),"--out",f"runs/Model1_Sentiment_DistilBERT_{seed}"]
    subprocess.run(command,check=True)
```
